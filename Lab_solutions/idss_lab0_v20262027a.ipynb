# Part 1 &mdash; Vectors


## Task 1 &mdash; Make a vector &nbsp;*(1 mark)*

`np.array([...])` turns a Python list of numbers into a NumPy vector.

Create the vector

$$\mathbf{a} = (2,\ -1,\ 4)$$

and store it in a variable called `a`.


```python
import numpy as np
a = np.array([2, -1, 4])
```

```python
# This cell is used for grading purposes. Do not modify it.

with tick.marks(1):
    assert a.shape == (3,)
    assert np.allclose(a, [2.0, -1.0, 4.0])
```

## Task 2 &mdash; Vector arithmetic &nbsp;*(2 marks)*

Vectors add **elementwise**, and multiplying by a scalar scales every element.

Using `a` from Task 1 and `b` defined in the next cell, compute

$$\mathbf{c} = 3\mathbf{a} - \mathbf{b}$$

and store the result in `c`.


```python
# provided data -- do not change
b = np.array([1.0, 1.0, 1.0])
```

```python
c = ((3 * a) - b)
```

```python
# This cell is used for grading purposes. Do not modify it.

with tick.marks(2):
    assert c.shape == (3,)
    assert np.allclose(c, [5.0, -4.0, 11.0])
```

## Task 3 &mdash; Inner (dot) product &nbsp;*(2 marks with feedback + 2 marks hidden)*

The **inner product** of two vectors is the sum of their elementwise products,
$\;\mathbf{u}\cdot\mathbf{v} = \sum_i u_i v_i$. In NumPy this is `u @ v` (or `np.dot(u, v)`).

Write a function `inner(u, v)` that returns the inner product of two vectors `u` and `v`.

*Why a function?* The hidden test calls `inner` with a different pair of vectors, so you cannot
pass by hard-coding the answer &mdash; the assessed labs work the same way.


```python
def inner(u, v):
    return np.dot(u, v)
```

```python
# This cell is used for grading purposes. Do not modify it.

with tick.marks(2):
    assert np.isclose(inner(np.array([1.0, 2.0, 3.0]), np.array([4.0, 5.0, 6.0])), 32.0)
```

## Task 4 &mdash; Length of a vector &nbsp;*(2 marks)*

The **Euclidean norm** (or L2 norm) $\;\|\mathbf{a}\|_2 = \sqrt{\sum_i a_i^2}\;$ measures the length
of a vector. Use `np.linalg.norm`.

Store the norm of `a` in `a_len`.


```python
a_len = np.linalg.norm(a)
```

```python
# This cell is used for grading purposes. Do not modify it.

with tick.marks(2):
    assert np.isclose(a_len, np.sqrt(21.0))   # sqrt(2^2 + 1^2 + 4^2)
```

## Task 7 &mdash; The mean vector &nbsp;*(1 mark with feedback + 2 marks hidden)*

Data is often stored as a matrix with **one row per observation**. The **mean vector** is the
average of the rows &mdash; the "centre of mass" of the data.

`D` below is a 4&times;3 dataset. Compute its mean vector (the average over the rows) and store it in
`mean_vec`.

*Hint:* `D.mean(axis=0)` averages down the columns, giving one value per feature.


```python
# provided data -- do not change
D = np.array([
    [2.0,  4.0,  6.0],
    [4.0,  8.0, 12.0],
    [0.0,  0.0,  0.0],
    [6.0, 12.0, 18.0],
])
```

```python
mean_vec = D.mean(axis=0)
```
