# Part 1: Eigenvalues and eigenvectors


## Task 1: Eigenvalues and eigenvectors &nbsp;*(1 mark with feedback + 2 marks hidden)*

For a square matrix $M$, a non-zero vector $v$ is an **eigenvector** of $M$ if multiplying by $M$
only scales $v$, without changing its direction:

$$Mv = \lambda v$$

The number $\lambda$ is the matching **eigenvalue**. A matrix of size $n \times n$ has $n$
eigenvalue and eigenvector pairs (counted with repeats).

In Lab 1, Task 8, you multiplied a starting vector by a matrix again and again, and the result moved
towards a fixed point. That fixed point is closely related to the eigenvector with the largest
eigenvalue, often called the **dominant** eigenvector. We will not repeat that calculation here,
because the matrix from Lab 1 is not symmetric, and its eigenvalues are not as easy to work with. In
this task you will practise on a matrix where everything stays simple: a **symmetric** matrix,
where $M$ equals its own transpose. For a symmetric matrix, NumPy's `np.linalg.eigh` function
returns real eigenvalues, listed from smallest to largest, and a matching matrix of eigenvectors, one
per column, each with length 1.

`M` below is a symmetric 3 by 3 matrix. Using `np.linalg.eigh`:

1. Store the array of eigenvalues in `eigvals` and the array of eigenvectors (one per column) in
   `eigvecs`.
2. Find the position of the largest eigenvalue. Store that eigenvalue in `dominant_eigval`, and its
   matching eigenvector (as a plain 1D array) in `dominant_eigvec`.

*Hint:* `np.argmax` finds the position of the largest value in an array. `eigvecs` stores each
eigenvector as a **column**, not a row, so the eigenvector matching `eigvals[k]` is the column
`eigvecs[:, k]`, not the row `eigvecs[k]`. These generally contain different numbers, so picking the
row instead of the column is a common and easy mistake to make.


```python
# provided data, do not change
M = np.array([
    [2.0, 1.0, 0.0],
    [1.0, 2.0, 1.0],
    [0.0, 1.0, 2.0],
])
```

```python
eigvals, eigvecs = np.linalg.eigh(M)
i = np.argmax(np.abs(eigvals))
dominant_eigval = eigvals[i]
dominant_eigvec = eigvecs[:, i]
```

```python
with tick.marks(1):
    assert eigvals.shape == (3,)
    assert eigvecs.shape == (3, 3)
    assert dominant_eigvec.shape == (3,)
    # the defining property of an eigenpair, true for any correct answer:
    assert np.allclose(M @ dominant_eigvec, dominant_eigval * dominant_eigvec, atol=1e-6), (
        "M @ dominant_eigvec should equal dominant_eigval * dominant_eigvec. "
        "Check that dominant_eigvec is a COLUMN of eigvecs (eigvecs[:, k]), not a row (eigvecs[k])."
    )
```

# Part 2: Principal component analysis


## Task 2: Mean-centring and the covariance matrix &nbsp;*(2 marks with feedback + 1 mark hidden)*

Principal component analysis (PCA) finds the directions in a dataset along which the data varies the
most. The first step, as in Lab 1, is to centre the data by subtracting the mean vector.

The second step is to build the **covariance matrix** of the centred data. For a dataset with $N$
rows (observations), stored as a matrix $X_c$ that has already been centred, the covariance matrix is

$$\Sigma = \frac{1}{N - 1} X_c^T X_c$$

Each entry $\Sigma_{ij}$ measures how much feature $i$ and feature $j$ vary together. $\Sigma$ is
always square and symmetric, whatever the shape of $X_c$.

`X` below is a dataset with 6 rows and 2 columns.

1. Compute its mean vector and store it in `mean_vec`.
2. Subtract the mean vector from `X` to get the centred data, and store it in `X_centered`.
3. Compute the covariance matrix of `X_centered`, using the formula above, and store it in `cov`.

*Hint:* every piece of the formula above is something you already know how to write in NumPy.


```python
# provided data, do not change
X = np.array([
    [1.0, 2.0],
    [2.0, 3.0],
    [3.0, 3.5],
    [4.0, 5.5],
    [5.0, 5.0],
    [6.0, 7.0],
])
N = X.shape[0]
```

```python
mean_vec = np.mean(X, axis=0)
X_centered = X - mean_vec
cov = (X_centered.T @ X_centered) / (N - 1)
```

```python
with tick.marks(2):
    assert mean_vec.shape == (2,)
    assert X_centered.shape == (6, 2)
    assert cov.shape == (2, 2)
    assert np.allclose(X_centered.mean(axis=0), 0.0, atol=1e-8)
    assert np.allclose(cov, cov.T, atol=1e-8)  # covariance matrices are symmetric
    # the diagonal of a covariance matrix is the variance of each feature
    assert np.isclose(cov[0, 0], X_centered[:, 0].var(ddof=1), atol=1e-6)
    assert np.isclose(cov[1, 1], X_centered[:, 1].var(ddof=1), atol=1e-6)
```

## Task 3: Principal components from the covariance matrix &nbsp;*(1 mark with feedback + 2 marks
hidden)*

The covariance matrix `cov` is square and symmetric, so it has a full set of real eigenvalues and
eigenvectors, exactly as in Task 1. Its eigenvectors are the dataset's **principal components**: the
directions of greatest variation. The matching eigenvalue tells you how much variance lies along
that direction. This amount of variance is called the **explained variance** of that component.

1. Find the eigenvalues and eigenvectors of `cov`, using the same NumPy function as in Task 1. Store
   them in `cov_eigvals` and `cov_eigvecs`.
2. Identify the **top** principal component: the eigenvector with the largest eigenvalue. Store that
   eigenvector in `pc1`, and its eigenvalue (the explained variance of the top component) in
   `pc1_val`.


```python
cov_eigvals, cov_eigvecs = np.linalg.eigh(cov)
i = np.argmax(np.abs(cov_eigvals))
pc1 = cov_eigvecs[:, i]
pc1_val = cov_eigvals[i]
```

```python
with tick.marks(1):
    assert cov_eigvals.shape == (2,)
    assert cov_eigvecs.shape == (2, 2)
    assert pc1.shape == (2,)
    assert np.isclose(pc1_val, cov_eigvals.max())
    # the defining property of an eigenpair, true for any correct answer:
    assert np.allclose(cov @ pc1, pc1_val * pc1, atol=1e-6)
```

## Task 4: Projecting onto the top principal component &nbsp;*(1 mark with feedback + 1 mark
hidden)*

Once you have a principal component, you can **project** each centred data point onto it. This gives
one number per row: how far along that direction the point sits. For a centred data matrix $X_c$ and
a principal component $v$ (a unit-length column vector), the projected values are

$$\text{proj} = X_c v$$

This is the first step of dimensionality reduction: instead of describing each point with 2 numbers
(its original features), you describe it with 1 number (its position along the top component).

Using `X_centered` from Task 2 and `pc1` from Task 3, compute the projection of every row onto `pc1`,
and store the result in `proj`.


```python
proj = X_centered @ pc1
```

```python
with tick.marks(1):
    assert proj.shape == (6,)
    # the variance of the projection equals the explained variance of the component
    assert np.isclose(proj.var(ddof=1), pc1_val, atol=1e-6)
```

# Part 3: The same components, found with the singular value decomposition


## Task 5: The SVD of the centred data &nbsp;*(2 marks with feedback + 1 mark hidden)*

In Part 2 you found the principal components by building a square covariance matrix and
eigendecomposing it. There is a second route to the same answer, which works directly on the
original, non-square data matrix: the **singular value decomposition** (SVD).

For any matrix $A$ with $N$ rows and $p$ columns, the SVD writes it as

$$A = U S V^T$$

where $U$ is $N \times p$ with orthonormal columns, $V$ is $p \times p$ with orthonormal columns,
and $S$ is a 1D array of $p$ non-negative numbers, the **singular values**, listed from largest to
smallest.

Compute the SVD of `X_centered` (6 rows, 2 columns) using `np.linalg.svd`. Pass
`full_matrices=False`, so that `U` comes back with the same shape as `X_centered`, instead of a
larger, square matrix. Store the three results in `U`, `S` and `Vt` (NumPy returns $V^T$ directly,
not $V$).


```python
U, S, Vt = np.linalg.svd(X_centered, full_matrices=False)
```

```python
with tick.marks(2):
    assert U.shape == (6, 2)
    assert S.shape == (2,)
    assert Vt.shape == (2, 2)
    assert np.all(S[:-1] >= S[1:])  # singular values are listed largest first
    # putting the pieces back together should reproduce the centred data
    assert np.allclose(U @ np.diag(S) @ Vt, X_centered, atol=1e-6)
```

## Task 6: Checking that both routes agree &nbsp;*(1 mark with feedback + 2 marks hidden)*

The SVD and the covariance eigendecomposition should find the same directions. Two facts connect
them, for data that has already been centred:

* the rows of $V^T$ (the right singular vectors) are the same as the covariance matrix's
  eigenvectors, listed in the same order of importance;
* the singular values relate to the covariance eigenvalues (the explained variances) by
  $\lambda_i = \dfrac{\sigma_i^2}{N - 1}$.

1. Using `S` from Task 5, compute the explained variance of every component this way, and store the
   array in `svd_explained_variance`.
2. Compare the top row of `Vt` with `pc1` from Task 3. They should be equal, **or they should be
   equal after flipping the sign of one of them**: both an eigenvector and a singular vector point
   along a line, not in a single fixed direction, so either sign is a correct answer. Write a check
   for this and store the result, a single `True` or `False` value, in `pc1_matches_svd`.

*Hint:* `np.allclose(a, b)` tests whether two arrays are equal, within a small numerical tolerance.
You can test the flipped case with `np.allclose(a, -b)`. The default tolerance is very strict, and
values that come out of an eigendecomposition or an SVD carry a small amount of rounding error, so
the default can reject two vectors that are really the same. Pass a looser absolute tolerance with
the `atol` argument, for example `np.allclose(a, b, atol=1e-6)`, which accepts a difference of up
to $10^{-6}$ in each entry.


```python
svd_explained_variance = S**2 / (N - 1)
pc1_matches_svd = np.allclose(Vt[0], pc1, atol=1e-6) or np.allclose(Vt[0], -pc1, atol=1e-6)
```

```python
with tick.marks(1):
    assert svd_explained_variance.shape == (2,)
    assert bool(pc1_matches_svd) is True
    # the two routes to PCA should agree on how much variance each component explains
    assert np.allclose(sorted(svd_explained_variance), sorted(cov_eigvals), atol=1e-6)
```

# Part 4: Using the SVD to approximate data


## Task 7: A rank-1 approximation of the data &nbsp;*(2 marks with feedback + 1 mark hidden)*

Because the singular values are listed largest first, you can build an approximate version of a
dataset by keeping only the first few terms of the SVD and dropping the rest. 
Keeping only the single largest singular value and its matching vectors gives a **rank-1 approximation**: the
closest possible dataset that varies along only one direction. 

Using the first column $u_1$ of $U$,
the first singular value $\sigma_1$, and the first row $v_1^T$ of $V^T$, the rank-1 approximation
of the centred data is

$$X_c^{(1)} = \sigma_1 \, u_1 v_1^T$$

Another way to see this: it is the same as reconstructing the data from the *full* SVD,
$X_c = U \, \mathrm{diag}(S) \, V^T$, but with every singular value except $\sigma_1$ replaced by
zero. A zeroed singular value makes its whole term vanish from the sum, so the matching column of
`U` and row of `V^T` stop contributing. This is why the result still has the same shape as the
original data: nothing is deleted, the unneeded directions are simply multiplied by zero.

1. Build this approximation, then
2. add the mean vector back on, storing the result (the same shape as
the original `X`) in `X_reconstructed`. 

The final line in the answer cell is provided, and it
computes how far this approximation is from the real data.

*Hint:* slicing the first column of a 2D array, for example `U[:, :1]`, keeps the result 2D rather
than flattening it to 1D. Remember that `S` is a plain list of numbers, not yet a matrix: you will
need to turn the part of it you keep into a diagonal matrix before you can multiply by it.

*Hint on `np.diag`:* given a 1D array, `np.diag` builds a 2D diagonal matrix from it, placing the
values along the diagonal and filling everywhere else with 0. For example, `np.diag(np.array([5.0]))`
gives the 1 by 1 matrix `[[5.0]]`. (Given a 2D array instead, `np.diag` does the opposite: it reads
off the diagonal values.) You want $\sigma_1$ alone turned into a matrix, so apply `np.diag` to the
sliced `S[:1]`, not to the whole `S`.

The shapes you are aiming for, for this dataset's `X` (6 rows, 2 features):

| variable | how you might build it | shape |
|---|---|---|
| `u1` | `U[:, :1]` | (6, 1) |
| `s1` | `np.diag(S[:1])` | (1, 1) |
| `v1` | `Vt[:1, :]` | (1, 2) |


```python
X_reconstructed = (U[:, :1] @ np.diag(S[:1]) @ Vt[:1, :]) + mean_vec
# provided: how far the rank-1 approximation is from the real data
reconstruction_error = np.linalg.norm(X - X_reconstructed)
```

```python
with tick.marks(2):
    assert X_reconstructed.shape == (6, 2)
    assert reconstruction_error > 0  # the rank-1 approximation is not a perfect copy
    # a genuine rank-1 approximation varies along only one direction
    assert np.linalg.matrix_rank(X_reconstructed - mean_vec, tol=1e-6) == 1
```

## Beyond this lab

This is the last lab in the linear algebra strand, but Lecture 2 continues past the SVD, into
matrix inversion, linear systems, rank and condition number, and a method called whitening. There is
no task on these here, but remember to revisit the following:

* **Matrix inversion**: for a square matrix $A$, its inverse $A^{-1}$ satisfies $A^{-1} A = I$. It
  is the matrix version of "undo". Only square matrices can have an inverse, and even some square
  matrices, called **singular** matrices, do not have one.
* **Linear systems**: an equation of the form $Ax = b$ is a linear system. If $A$ has an inverse, the
  solution is $x = A^{-1} b$. In practice, `np.linalg.solve` finds $x$ directly. It is faster and
  more numerically stable than computing the inverse first.
* **Rank**: the rank of a matrix is the number of independent directions it uses. A matrix with full
  rank keeps all of its information. A matrix with low rank, such as the rank-1 approximation you
  built in Task 7, has thrown some information away on purpose.
* **Condition number**: a single number that tells you how sensitive the solution of a linear system
  is to small changes, or small errors, in the input. A very large condition number is a warning
  sign.
* **Whitening**: a transformation, built from an eigendecomposition or an SVD, that rescales a
  dataset so that every direction has the same variance. It is a common preprocessing step before
  other algorithms.
